In this example, we're given a task to take the raw dataset to clean & preprocess. In this LOAN-DATA CSV file we will:
1) Examine the Data
2) Import the Data
3) Split the Data

NOTE: THE DATASET WE ARE USING IS FROM A BANK IN U.S WITH CURRENCY AS $.
EVERY CATEGORICAL VARIABLE MUST BE QUANTIFIED, SO WE NEED TO CHANGE ANY TEXT COLUMNS INTO NUMBERS.
NOW, WHEN WE'RE MEASURING CREDITWORTHINESS:
1. WE NEED TO BE EXTREMELY RISK-AVERSE AND DISTRUSTFUL.
2. MISSING INFORMATION SUGGESTS FOUL PLAY.
3. IF THE INFORMATION ISN'T AVAILABLE, WE'LL ASSUME THE WORST.

```python
#importing dataset via loadtxt we got an error due to NAN which could be missing values or text so we must use genfromtxt and can add skip header for first row as its all NAN.
raw_data_np = np.genfromtxt("loan-data.csv", delimiter = ';', skip_header = 1, autostrip = True)
raw_data_np
```

```python
#Checking for missing values
np.isnan(raw_data_np).sum()
```

```python
#Let's create two temporary variables for missing values
temporary_fill = np.nanmax(raw_data_np) + 1
temporary_mean = np.nanmean(raw_data_np, axis = 0)
```

```python
temporary_mean
```

```python
#We should split numeric and strings in different arrays, before we do that we should determine the min and max values for future substitution.
temporary_stats = np.array([np.nanmin(raw_data_np, axis = 0), 
                            temporary_mean, 
                            np.nanmax(raw_data_np, axis = 0)])
```

```python
temporary_stats
```

```python
#Splitting up the data: to recall the isnan function returns true if non numeric is present with the np.argwhere function. no by applying by this strategy we get the indices of strings.
column_strings = np.argwhere(np.isnan(temporary_mean)).squeeze()
column_strings
#we used .squeeze method at the end to display it in one line dimension
```

```python
#Now to also know columns that contain numeric values we'll add == false to our code
column_numeric = np.argwhere(np.isnan(temporary_mean) == False).squeeze()
column_numeric
```

```python
#Now by knowing above indices. we can reimport our dataset with : usecols = columns_strings, columns_numeric
loan_data_strings = np.genfromtxt("loan-data.csv",
                                  delimiter = ';',
                                  skip_header = 1,
                                  autostrip = True,
                                  usecols = column_strings,
                                  dtype = str)
loan_data_strings
```

```python
#Now for missing value shown before 'A1' in Subgrade we can use grade for more appropriate estimation.
#For example if there's a missing value in subgrade with a grade of 'B' we can assign 'B5' rather than the lowest 'G5'.
#By the help of iteration. In the first line we define the loop which goes through all the grades after the first one'[:,3][1:]'
#which is the empty space. Then for every grade in line two we go through column to check if the subgrade is missing by the np.where function.
#If the subgrade turns out to be missing then we would want to also check whether the grade for that same row equals the grade for the same iteration.
#once that is also satisfied we'll assign the worst(lowest) subgrade (i + 5) otherwise columns should remain unchanged(last line)
for i in np.unique(loan_data_strings[:,3])[1:]:
    loan_data_strings[:,4] = np.where((loan_data_strings[:,4] == '') & (loan_data_strings[:,3] == i),
                                      i + '5',
                                      loan_data_strings[:,4])
```

```python
#Based in European office we need exchange rates between Euro USD at the time of loan application.
#EUR-USD.csv file contains all exchange rates of 2015
EUR_USD = np.genfromtxt("EUR-USD.csv", delimiter = ',', autostrip = True, skip_header = 1, usecols = 3)
EUR_USD
```

```python
loan_data_strings[:,0]
```

```python
#Now we'll create a for loop that goes through the range 1-13, 1 iteration for every month of the year because the upper limit 13 is not included.
#We use [i-1] here because indexing in python starts with 0. 
exchange_rate = loan_data_strings[:,0]

for i in range(1,13):
    exchange_rate = np.where(exchange_rate == i,
                             EUR_USD[i-1],
                             exchange_rate)

#For accounts with no issue dates, we're going to use the annual exchange rate of 2015.

exchange_rate = np.where(exchange_rate == 0,
                             np.mean(EUR_USD),
                             exchange_rate)
exchange_rate
```

```python
#Now lets add the exchange rates to our dataset it to our Dataset.
#Checking for shape compatibilty of both
exchange_rate.shape
```

```python
loan_data_numeric.shape
```

```python
#As exchange_rate is one dimentional and loan_data_numeric is two dimentional. We must reshape the new array.
exchange_rate = np.reshape(exchange_rate, (10000,1))
```

```python
#Now let's horizontally add the new array into our dataset as both are two dimentional
loan_data_numeric = np.hstack((loan_data_numeric, exchange_rate))
```

```python
loan_data_numeric
```

```python
#Since we added a column to the array, we'll also include it to our header_numeric.
#Without adding np.array before 'exchange_rate' we would get an error because the header variable is one dimentional while the text 'exchange_rate' is a zero dimentional scaler.
#We'll change the text into an array with np.array to solve this issue.
header_numeric = np.concatenate((header_numeric, np.array(['exchange_rate'])))
header_numeric
```
